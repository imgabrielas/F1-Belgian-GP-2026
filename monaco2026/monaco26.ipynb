```python
import fastf1
import pandas as pd

fastf1.Cache.enable_cache("cache")

YEAR = 2026
EVENT = "Monaco"
```

```python
practice_frames = []

for session_name in ["FP1", "FP2", "FP3"]:

    session = fastf1.get_session(YEAR, EVENT, session_name)
    session.load()

    df = session.results.copy()
    df["Session"] = session_name

    practice_frames.append(df)

practice_df = pd.concat(practice_frames, ignore_index=True)
```

```python
qualifying = fastf1.get_session(YEAR, EVENT, "Q")
qualifying.load()

qualifying_df = qualifying.results.copy()
```

```python
race = fastf1.get_session(YEAR, EVENT, "R")
race.load()

race_df = race.results.copy()
```
