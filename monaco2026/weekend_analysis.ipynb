### Qualification

```python
import pandas as pd

df = pd.read_csv('weekend_data/qualifying.csv')
df
```

```python
df = df.drop(columns=['BroadcastName','DriverId','TeamColor','TeamId','FirstName','LastName','FullName','HeadshotUrl','CountryCode','ClassifiedPosition','GridPosition','Time','Status','Points','Laps'], errors='ignore')

df
```

```python
q1_td = pd.to_timedelta(df['Q1'], errors='coerce')
top7 = df.loc[q1_td.nlargest(7).index]
print(f"Drivers who finished their qualifying on Q1: {', '.join(top7['Abbreviation'])}")
```

This seems interesting LIN has worse time than BOR, yet LIN goes to Q2, did BOR DNF?

```python
q2_td = pd.to_timedelta(df['Q2'], errors='coerce')
top5 = df.loc[q2_td.nlargest(5).index]
print(f"Drivers who finished their qualifying on Q2: {', '.join(top5['Abbreviation'])}")
```

```python
remaining_q1 = df['Abbreviation'].head(10)
print(", ".join(remaining_q1))
```

```python
fastest_idx = q1_td.idxmin()
slowest_idx = q1_td.idxmax()

fastest_driver = df.loc[fastest_idx, 'Abbreviation']
slowest_driver = df.loc[slowest_idx, 'Abbreviation']
best_time = q1_td.min()
worst_time = q1_td.max()
gap = worst_time - best_time

print(f"Fastest Q1 time: {fastest_driver} - {best_time}")
print(f"Slowest Q1 time: {slowest_driver} - {worst_time}")
print(f"Gap between fastest and slowest: {gap.total_seconds():.3f} seconds")
print(f"On a track as tight as Monaco, a gap of {gap.total_seconds():.3f}s over a ~72s lap is huge") 
```

```python
q3_td = pd.to_timedelta(df['Q3'], errors='coerce')
fastest_idx_q3 = q3_td.idxmin()
slowest_idx_q3 = q3_td.idxmax()

fastest_driver_q3 = df.loc[fastest_idx_q3, 'Abbreviation']
slowest_driver_q3 = df.loc[slowest_idx_q3, 'Abbreviation']
best_q3_time = q3_td.min()
worst_q3_time = q3_td.max()
gap_q3 = worst_q3_time - best_q3_time

print(f"Fastest Q3 time: {fastest_driver_q3} - {best_q3_time}")
print(f"Slowest Q3 time: {slowest_driver_q3} - {worst_q3_time}")
print(f"Gap between fastest and slowest in Q3: {gap_q3.total_seconds():.3f} seconds")
slowest_driverQ3 = df.loc[slowest_idx, 'Abbreviation']

print("")

print(f"Fastest Q1 time: {fastest_driver} - {best_time}")
print(f"Slowest Q3 time: {slowest_driverQ3} - {worst_time}")
print(f"Gap between fastest and slowest: {gap.total_seconds():.3f} seconds")
```

```python
improvement = pd.DataFrame({
    'Abbreviation': df['Abbreviation'],
    'Q1': q1_td.dt.total_seconds(),
    'Q2': q2_td.dt.total_seconds(),
    'Q3': q3_td.dt.total_seconds(),
})

# Positive = got faster (improved), negative = got slower
improvement['Q1_to_Q2'] = improvement['Q1'] - improvement['Q2']
improvement['Q2_to_Q3'] = improvement['Q2'] - improvement['Q3']

q1_to_q2 = improvement.dropna(subset=['Q1_to_Q2']).sort_values('Q1_to_Q2', ascending=False)
q2_to_q3 = improvement.dropna(subset=['Q2_to_Q3']).sort_values('Q2_to_Q3', ascending=False)

print("Q1 -> Q2 improvement (seconds gained), only drivers who advanced:")
print(q1_to_q2[['Abbreviation', 'Q1', 'Q2', 'Q1_to_Q2']].to_string(index=False))

print("\nQ2 -> Q3 improvement (seconds gained), only drivers who advanced:")
print(q2_to_q3[['Abbreviation', 'Q2', 'Q3', 'Q2_to_Q3']].to_string(index=False))

biggest_q1q2 = q1_to_q2.iloc[0]
biggest_q2q3 = q2_to_q3.iloc[0]
print(f"\nBiggest Q1->Q2 gain: {biggest_q1q2['Abbreviation']} found {biggest_q1q2['Q1_to_Q2']:.3f}s")
print(f"Biggest Q2->Q3 gain: {biggest_q2q3['Abbreviation']} found {biggest_q2q3['Q2_to_Q3']:.3f}s")
```
