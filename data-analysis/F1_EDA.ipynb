```python
import fastf1
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

import matplotlib.pyplot as plt

# fastf1.Cache.enable_cache("cache")
```

```python
YEAR = 2026
```

```python
schedule = fastf1.get_event_schedule(YEAR)

schedule
```

```python
fig = px.scatter(
    schedule,
    x="RoundNumber",
    y="Country",
    hover_name="EventName",
    title=f"{YEAR} Formula 1 Calendar"
)

fig.show()
```

```python
race = fastf1.get_session(YEAR, "Monaco", "R")
race.load()
```

```python
race.results
```

```python
TEAM_COLORS = {
    "Mercedes": "#00D2BE",
    "Ferrari": "#DC0000",
    "McLaren": "#FF8700",
    "Red Bull Racing": "#1E41FF",
    "Alpine": "#0090FF",
    "Racing Bulls": "#2B45FF",
    "Haas F1 Team": "#FFFFFF",
    "Williams": "#005AFF",
    "Audi": "#E30613",
    "Aston Martin": "#006F62",
    "Cadillac": "#0A0A0A"
}

TEAM_DRIVERS = {
    "Mercedes": ["RUS", "ANT"],
    "Ferrari": ["LEC", "HAM"],
    "McLaren": ["NOR", "PIA"],
    "Red Bull Racing": ["VER", "TSU"],
    "Alpine": ["COL", "GAS"],
    "Racing Bulls": ["HAD", "LAW"],
    "Haas F1 Team": ["OCO", "BEA"],
    "Williams": ["ALB", "SAI"],
    "Audi": ["HUL", "BOR"],
    "Aston Martin": ["ALO", "STR"],
    "Cadillac": ["PER", "ZHO"],
}

DRIVER_COLORS = {
    driver: TEAM_COLORS[team]
    for team, drivers in TEAM_DRIVERS.items()
    for driver in drivers
}
```

```python
fig = px.bar(
    race.results,
    x="Abbreviation",
    y="Points",
    color="TeamName",
    color_discrete_map=TEAM_COLORS,
    title="Race Points"
)

fig.show()
```

```python
race.results["Places Gained"] = (
    race.results["GridPosition"] - race.results["Position"]
)

fig = px.bar(
    race.results.sort_values("Places Gained"),
    x="Abbreviation",
    y="Places Gained",
    color="TeamName",
    color_discrete_map=TEAM_COLORS,
    title="Places Gained During Race"
)

fig.show()
```

```python
all_results = []

for event in schedule["EventName"]:

    session = fastf1.get_session(YEAR, event, "R")
    session.load()

    df = session.results.copy()

    df["Race"] = event

    all_results.append(df)
```

```python
results_df = pd.concat(all_results)
```

```python
points = (
    results_df
    .groupby(["Race", "Abbreviation"])["Points"]
    .sum()
    .reset_index()
)

fig = px.line(
    points,
    x="Race",
    y="Points",
    color="Abbreviation",
    color_discrete_map=DRIVER_COLORS,
    markers=True,
    title="Points Scored per Grand Prix"
)

fig.update_layout(
    xaxis_title="Grand Prix",
    yaxis_title="Points",
    legend_title="Driver",
    template="plotly_dark",
    hovermode="x unified"
)

fig.update_xaxes(tickangle=-45)

fig.show()
```

```python
team_points = (
    results_df
    .groupby("TeamName")["Points"]
    .sum()
    .sort_values(ascending=False)
)

team_points = team_points.reset_index()

fig = px.bar(
    team_points,
    x="Points",
    y="TeamName",
    color="TeamName",
    color_discrete_map=TEAM_COLORS,
    orientation="h",
    title="Constructor Championship Points"
)

fig.update_layout(
    xaxis_title="Points",
    yaxis_title="Constructor",
    template="plotly_dark",
    showlegend=False
)

fig.update_yaxes(categoryorder="total ascending")

fig.show()
```

```python
results_df["Places Gained"] = (
    results_df["GridPosition"] - results_df["Position"]
)

places = (
    results_df
    .groupby("Abbreviation", as_index=False)["Places Gained"]
    .mean()
    .sort_values("Places Gained", ascending=False)
)
```

```python
podiums = (
    results_df[results_df["Position"] <= 3]
    .groupby("Abbreviation", as_index=False)
    .size()
    .rename(columns={"size": "Podiums"})
    .sort_values("Podiums", ascending=False)
)

fig = px.bar(
    podiums,
    x="Abbreviation",
    y="Podiums",
    color="Abbreviation",
    color_discrete_map=DRIVER_COLORS,
    title="Podiums"
)

fig.update_layout(
    template="plotly_dark",
    showlegend=False
)

fig.show()
```

```python
dnfs = (
    results_df[results_df["Status"] != "Finished"]
    .groupby("Abbreviation", as_index=False)
    .size()
    .rename(columns={"size": "DNFs"})
    .sort_values("DNFs", ascending=False)
)

fig = px.bar(
    dnfs,
    x="Abbreviation",
    y="DNFs",
    color="Abbreviation",
    color_discrete_map=DRIVER_COLORS,
    title="DNFs"
)

fig.update_layout(
    template="plotly_dark",
    showlegend=False
)

fig.show()
```

```python
comparison = (
    results_df
    .groupby("Abbreviation", as_index=False)
    .agg({
        "GridPosition": "mean",
        "Position": "mean"
    })
)

fig = px.scatter(
    comparison,
    x="GridPosition",
    y="Position",
    color="Abbreviation",
    color_discrete_map=DRIVER_COLORS,
    text="Abbreviation",
    title="Qualifying vs Race Performance"
)

fig.update_traces(textposition="top center")

fig.update_layout(
    template="plotly_dark",
    xaxis_title="Average Grid Position",
    yaxis_title="Average Finishing Position",
    showlegend=False
)

fig.update_xaxes(autorange="reversed")
fig.update_yaxes(autorange="reversed")

fig.show()
```
